```python
import numpy as np
import matplotlib.pyplot as plt

T_0 = 100
T_inf = 20
a = 1


def T(a, x, y, T_0, T_inf):
    return T_0 + (T_inf - T_0) * (1 - a / (np.sqrt(x**2 + y**2)))


N = 100
x = np.linspace(-5, 5, N)
y = np.linspace(-2, 2, N)
X, Y = np.meshgrid(x, y)

T_distr = T(a, X, Y, T_0, T_inf)
T_distr[X**2 + Y**2 <= a**2] = T_0

# Parte 1, campo de temperatura

fig, ax = plt.subplots()
contour = ax.contourf(X, Y, T_distr)
c = fig.colorbar(contour)
ax.set_aspect('equal')
fig.show()
```

```python
k = 5
q = np.gradient(T_distr) * np.ones_like(T_distr) * -k

fig, ax = plt.subplots(figsize=(10, 5))
contour = ax.contourf(X, Y, T_distr)
ax.quiver(X, Y, q[1], q[0])
c = fig.colorbar(contour)
ax.set_aspect('equal')
fig.show()
```

## Problema 2

```python
N = 400
sY = 70

x = np.linspace(-5, 5, N)
y = np.linspace(-5, 5, N)

X, Y = np.meshgrid(x, y)

s_xx = -2 * X
s_yy = 2 * X
s_xy = 2*X * Y

S_vm = np.sqrt(s_xx**2 + s_yy**2 - s_xx*s_yy + 3*s_xy**2)

bool_svm = S_vm >= sY
S_vm[bool_svm] = np.nan

fig, ax = plt.subplots(figsize=(10, 10))
c = ax.contourf(X, Y, S_vm, levels=100, cmap='viridis')
fig.colorbar(c, ax=ax)
ax.scatter(X[bool_svm], Y[bool_svm], color="red")
ax.set_aspect('equal')
plt.show()
```

## Problema 3

```python
N = 400

x = np.linspace(-10, 10, N)
y = np.linspace(-5, 5, N)

X, Y = np.meshgrid(x, y)

Ux = Y
Uy = X
Uv = np.sqrt(Ux**2 + Uy**2)

fig, ax = plt.subplots()
c = ax.contourf((X+Ux), (Y+Uy), Uv, levels=100, cmap='viridis')
ax.set_aspect('equal')
plt.show()
```
