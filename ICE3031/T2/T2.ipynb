## Deduccion con sympy

```python
from sympy import Matrix, sin, cos, sqrt, log, pi
import sympy as sp

X1 = sp.symbols('X_1', real=True, nonnegative=True)
X2, X3 = sp.symbols('X_2 X_3', real=True,    nonnegative=True)
a, L = sp.symbols('alpha L', positive=True, real=True)


theta = a*X1/L * 3*pi/4
#theta = sp.Function('theta', real=True, nonnegative=True)(X1)

r = sqrt(1 + a*X1/L)
f1 = L/a * log(1 + a*X1/L)
f2 = r * (X2*cos(theta) - X3*sin(theta))
f3 = r * (X2*sin(theta) + X3*cos(theta))

phi = Matrix([f1, f2, f3])
X = Matrix([[X1], [X2], [X3]])

F = phi.jacobian(X)
F
```

```python
C = F.T @ F
C
```

```python
C = sp.trigsimp(C)
C
```

```python
E = 1/2 * (C - sp.eye(3))
E
```

```python
Finv = sp.trigsimp(F.inv())
Finv
```

```python
sp.trigsimp(Finv.T * Finv)
```

## Preguntas de Paraview

```python
import numpy as np
import pyvista as pv
```

```python
L = 5
h = 1
alpha = 1
Nx, Ny, Nz = 20, 4, 4

_X = np.linspace(0, L, Nx+1)
_Y = np.linspace(-h/2, h/2, Ny+1)
_Z = np.linspace(-h/2, h/2, Nz+1)

X, Y, Z = np.meshgrid(_X, _Y, _Z, indexing='ij')
```

```python
def phi(X1, X2, X3):
    r2 = 1 + alpha*X1/L
    theta = alpha * 3 * np.pi * X1 / (4*L)

    x1 = L/alpha * np.log(r2)
    x2 = np.sqrt(r2) * (X2*np.cos(theta) - X3*np.sin(theta))
    x3 = np.sqrt(r2) * (X2*np.sin(theta) + X3*np.cos(theta))
    return (x1, x2, x3)


def F(X1, X2, X3):
    r2 = 1 + alpha*X1/L
    theta = alpha * 3 * np.pi * X1 / (4*L)

    F11 = 1 / r2
    F12 = np.zeros_like(X1)
    F13 = np.zeros_like(X1)
    F21 = np.sqrt(r2)*(alpha*3*np.pi/(4*L)) * (-X2*np.sin(theta) - X3*np.cos(theta)) + alpha/(2*L) * (X2*np.cos(theta) - X3*np.sin(theta)) / np.sqrt(r2)
    F22 = np.sqrt(r2) * np.cos(theta)
    F23 = -np.sqrt(r2) * np.sin(theta)
    F31 = np.sqrt(r2)*(alpha*3*np.pi/(4*L)) * (X2*np.cos(theta) - X3*np.sin(theta)) + alpha/(2*L) * (X2*np.sin(theta) + X3*np.cos(theta)) / np.sqrt(r2)
    F32 = -F23
    F33 = F22
    return np.array([[F11, F12, F13], [F21, F22, F23], [F31, F32, F33]])


def Finv(X1, X2, X3):
    r2 = 1 + alpha*X1/L
    theta = alpha * 3 * np.pi * X1 / (4*L)

    F11 = r2
    F12 = np.zeros_like(X1)
    F13 = np.zeros_like(X1)
    F21 = (alpha * (-2*L*X2 + 3*np.pi*L*X3 + 3*np.pi*X1*X3*alpha) / (4*L**2))
    F22 = (np.sqrt(L) * (L + X1*alpha)**(3/2) * np.cos(theta) / (L**2 + 2*L*X1*alpha + X1**2 * alpha**2))
    F23 = (np.sqrt(L) * (L + X1*alpha)**(3/2) * np.sin(theta) / (L**2 + 2*L*X1*alpha + X1**2 * alpha**2))
    F31 = -(alpha *(3*np.pi*L*X2 + 2*L*X3 + 3*np.pi*X1*X2*alpha) / (4*L**2))
    F32 = -np.sqrt(L) * np.sin(theta) / np.sqrt(L + X1 * alpha)
    F33 = np.sqrt(L) * np.cos(theta) / np.sqrt(L + X1 * alpha)
    return np.array([[F11, F12, F13], [F21, F22, F23], [F31, F32, F33]])


def C(X1, X2, X3):
    r2 = 1 + alpha*X1/L
    #theta = alpha * 3 * np.pi * X1 / (4*L)

    C11 = r2**(-2) + alpha**2 * (X2**2 + X3**2) / (16*L**2 * r2) * (9*np.pi**2 * r2 + 4)
    C12 = alpha/(4*L) * (2*X2 - 3*np.pi*X3 * r2)
    C13 = alpha/(4*L) * (2*X3 + 3*np.pi*X2 * r2)
    C22 = r2
    C23 = np.zeros_like(X1)
    return np.array([[C11, C12, C13], [C12, C22, C23], [C13, C23, C22]])


def e(x1, x2, x3):
    I = np.eye(3)[:, :, np.newaxis, np.newaxis, np.newaxis]
    _Finv = Finv(x1, x2, x3)
    _FinvT = np.swapaxes(_Finv, 0, 1)
    return 1/2 * (I - np.einsum('ijxyz,jkxyz->ikxyz', _FinvT, _Finv))

```

```python
x, y, z = phi(X, Y, Z)
C_val = C(X, Y, Z)

E_val = 1/2 * (C_val - np.eye(3)[:, :, np.newaxis, np.newaxis, np.newaxis])
e_val = e(X, Y, Z)
```

```python
def ravel_points(x, y, z):
    return np.column_stack([
        x.ravel(order="F"),
        y.ravel(order="F"),
        z.ravel(order="F")])


Puntos = ravel_points(X, Y, Z)
puntos = ravel_points(x, y, z)
U = puntos - Puntos

grid_mat = pv.StructuredGrid(X, Y, Z)
grid_spa = pv.StructuredGrid(x, y, z)

grid_spa.point_data["Disp"] = U
grid_spa.point_data["|Disp|"] = np.linalg.norm(U, axis=1)

for i in range(3):
    for j in range(3):
        grid_mat.point_data[f"C{i+1}{j+1}"] = C_val[i,j, :, :, :].ravel(order="F")
        grid_mat.point_data[f"E{i+1}{j+1}"] = E_val[i,j, :, :, :].ravel(order="F")
        grid_spa.point_data[f"e{i+1}{j+1}"] = e_val[i,j, :, :, :].ravel(order="F")


grid_mat.save("beam_mat.vtk")
grid_spa.save("beam_spa.vtk")
```
