```python
from sympy import Matrix, sin, cos, sqrt, log, pi
import sympy as sp

X1 = sp.symbols('X_1', real=True, nonnegative=True)
X2, X3 = sp.symbols('X_2 X_3', real=True)
r, mu, lam = sp.symbols('R mu lambda', real=True, nonnegative=True)

phi = Matrix([(X2 + r) * sin(X1/r), (X2 + r) * cos(X1/r)-r, X3])
X = Matrix([[X1], [X2], [X3]])

F = phi.jacobian(X)
FinvT = sp.trigsimp(F.inv().T)
J = sp.trigsimp(F.det())
```

```python
import numpy as np

L = 5
H = 1
R = 10
Mu = 1
Lam = 10

fun_phi = sp.lambdify([X1, X2, X3], phi.subs({r: R}), 'numpy')
fun_P11 = sp.lambdify([X1, X2, X3], P[0,0].subs({r: R, mu: Mu, lam: Lam}), 'numpy')
fun_P22 = sp.lambdify([X1, X2, X3], P[1,1].subs({r: R, mu: Mu, lam: Lam}), 'numpy')
fun_P12 = sp.lambdify([X1, X2, X3], P[0,1].subs({r: R, mu: Mu, lam: Lam}), 'numpy')
fun_s11 = sp.lambdify([X1, X2, X3], sigma[0,0].subs({r: R, mu: Mu, lam: Lam}), 'numpy')
fun_s22 = sp.lambdify([X1, X2, X3], sigma[1,1].subs({r: R, mu: Mu, lam: Lam}), 'numpy')
fun_s12 = sp.lambdify([X1, X2, X3], sigma[0,1].subs({r: R, mu: Mu, lam: Lam}), 'numpy')
fun_T = sp.lambdify([X1, X2, X3], T.subs({r: R, mu: Mu, lam: Lam}), 'numpy')
fun_t = sp.lambdify([X1, X2, X3], t.subs({r: R, mu: Mu, lam: Lam}), 'numpy')
```

```python
N = 10

_X_1 = np.linspace(0, L, N)
_X_2 = np.linspace(-H/2, H/2, N)
_X_3 = np.linspace(-H/2, H/2, N)

X_1, X_2, X_3 = np.meshgrid(_X_1, _X_2, _X_3, indexing='ij')

x_1, x_2, x_3 = fun_phi(X_1, X_2, X_3)[:,0]
P11 = fun_P11(X_1, X_2, X_3)
P22 = fun_P22(X_1, X_2, X_3)
P12 = fun_P12(X_1, X_2, X_3)
s11 = fun_s11(X_1, X_2, X_3)
s22 = fun_s22(X_1, X_2, X_3)
s12 = fun_s12(X_1, X_2, X_3)
```

```python
# export to vtu
def ravel_points(x, y, z):
    return np.column_stack([
        x.ravel(order="F"),
        y.ravel(order="F"),
        z.ravel(order="F")])


Puntos = ravel_points(X_1, X_2, X_3)
puntos = ravel_points(x_1, x_2, x_3)
U = puntos - Puntos


import pyvista as pv

grid_X = pv.StructuredGrid(X_1, X_2, X_3)

grid_X.point_data["Disp"] = U
grid_X.point_data['P11'] = P11.ravel(order='F')
grid_X.point_data['P22'] = P22.ravel(order='F')
grid_X.point_data['P12'] = P12.ravel(order='F')
grid_X.point_data['s11'] = s11.ravel(order='F')
grid_X.point_data['s22'] = s22.ravel(order='F')
grid_X.point_data['s12'] = s12.ravel(order='F')

grid_X.save("beam.vtk")
```
